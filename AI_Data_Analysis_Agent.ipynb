```python
def inspect_dataset():
    """Returns basic information about the dataset."""
    return {
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": df.columns.tolist(),
        "data_types": df.dtypes.astype(str).to_dict(),
        "missing_values": df.isnull().sum().to_dict()
    }
def calculate_statistics(column):
    """Calculates statistics for a numerical column."""
    if column not in df.columns:
        return f"Column '{column}' does not exist."
    if not pd.api.types.is_numeric_dtype(df[column]):
        return f"Column '{column}' is not numerical."
    return {
        "mean": df[column].mean(),
        "median": df[column].median(),
        "minimum": df[column].min(),
        "maximum": df[column].max(),
        "standard_deviation": df[column].std()
    }
def find_missing_values():
    """Finds missing values in the dataset."""
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    if len(missing) == 0:
        return "There are no missing values in the dataset."
    return missing.to_dict()
def get_unique_values(column):
    """Returns unique values for a categorical column."""
    if column not in df.columns:
        return f"Column '{column}' does not exist."
    return df[column].unique().tolist()
```

```python
def agent_decide(user_question):
    tools_description = """
Available tools:

1. inspect_dataset
   - Use when the user asks about the structure, columns, rows,
     data types, or general information about the dataset.

2. calculate_statistics
   - Use when the user asks for mean, median, minimum, maximum,
     standard deviation, or numerical statistics.

3. find_missing_values
   - Use when the user asks about missing or null values.

4. get_unique_values
   - Use when the user asks about categories or unique values.
"""

    prompt = f"""
You are an AI Data Analysis Agent.

You have access to the following tools:

{tools_description}

User question:
{user_question}

Decide which tool should be used.

Return your answer in exactly this format:

TOOL: tool_name
COLUMN: column_name

If a column is not required, write:
COLUMN: NONE
"""

    response = client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt
    )
    return response.output_text
```

```python
def run_agent(user_question):
    decision = agent_decide(user_question)
    print("Agent Decision:")
    print(decision)
    print("\n")
    decision_lower = decision.lower()
    # Tool 1
    if "inspect_dataset" in decision_lower:
        result = inspect_dataset()
    # Tool 2
    elif "calculate_statistics" in decision_lower:
        column = decision.split("COLUMN:")[1].strip()
        result = calculate_statistics(column)
    # Tool 3
    elif "find_missing_values" in decision_lower:
        result = find_missing_values()
    # Tool 4
    elif "get_unique_values" in decision_lower:
        column = decision.split("COLUMN:")[1].strip()
        result = get_unique_values(column)
    else:
        result = "The agent could not determine the appropriate tool."
    return result
```

```python
def run_agent(user_question):
    try:
        # 1. Ask Gemini to choose a tool
        decision = agent_decide(user_question)
        print("Agent Decision:")
        print(decision)
        print()
        decision_upper = decision.upper()
        # Extract columns
        if "COLUMNS:" in decision_upper:
            columns_text = decision_upper.split("COLUMNS:", 1)[1].strip()
        else:
            columns_text = "NONE"
        # 2. Plot Scatter
        if "PLOT_SCATTER" in decision_upper:
            if columns_text == "NONE":
                result = "Two columns are required for a scatter plot."
            else:
                columns = [
                    col.strip()
                    for col in columns_text.split(",")
                ]
                if len(columns) < 2:
                    result = "Two columns are required for a scatter plot."
                else:
                    matching_columns = []
                    for column in columns[:2]:
                        matches = [
                            real_col
                            for real_col in df.columns
                            if real_col.upper() == column
                        ]
                        if matches:
                            matching_columns.append(matches[0])
                    if len(matching_columns) == 2:
                        result = plot_scatter(
                            matching_columns[0],
                            matching_columns[1]
                        )
                    else:
                        result = "One or both columns do not exist."
        # 3. Calculate Correlation
        elif "CALCULATE_CORRELATION" in decision_upper:
            if columns_text == "NONE":
                result = "Two columns are required for correlation."
            else:
                columns = [
                    col.strip()
                    for col in columns_text.split(",")
                ]
                if len(columns) < 2:
                    result = "Two columns are required for correlation."
                else:
                    matching_columns = []
                    for column in columns[:2]:
                        matches = [
                            real_col
                            for real_col in df.columns
                            if real_col.upper() == column
                        ]
                        if matches:
                            matching_columns.append(matches[0])
                    if len(matching_columns) == 2:
                        result = calculate_correlation(
                            matching_columns[0],
                            matching_columns[1]
                        )
                    else:
                        result = "One or both columns do not exist."
        # 4. Plot Histogram
        elif "PLOT_HISTOGRAM" in decision_upper:
            if columns_text == "NONE":
                result = "No column was provided."
            else:
                column = columns_text.split(",")[0].strip()
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = plot_histogram(
                        matching_columns[0]
                    )
                else:
                    result = f"Column '{column}' does not exist."
        # 5. Plot Bar Chart
        elif "PLOT_BAR_CHART" in decision_upper:
            if columns_text == "NONE":
                result = "No column was provided."
            else:
                column = columns_text.split(",")[0].strip()
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = plot_bar_chart(
                        matching_columns[0]
                    )
                else:
                    result = f"Column '{column}' does not exist."
        # 6. Inspect Dataset
        elif "INSPECT_DATASET" in decision_upper:
            result = inspect_dataset()
        # 7. Multiple Statistics
        elif "CALCULATE_MULTIPLE_STATISTICS" in decision_upper:
            if columns_text == "NONE":
                result = "No columns were provided."

            else:
                columns = [
                    col.strip()
                    for col in columns_text.split(",")
                ]
                actual_columns = []
                for col in columns:
                    matches = [
                        real_col
                        for real_col in df.columns
                        if real_col.upper() == col
                    ]
                    if matches:
                        actual_columns.append(matches[0])
                if len(actual_columns) == 0:
                    result = "None of the requested columns exist."
                else:
                    result = calculate_multiple_statistics(
                        actual_columns
                    )
        # 8. Single Statistics
        elif "CALCULATE_STATISTICS" in decision_upper:
            if columns_text == "NONE":
                result = "No column was provided."
            else:
                column = columns_text.split(",")[0].strip()
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = calculate_statistics(
                        matching_columns[0]
                    )
                else:
                    result = f"Column '{column}' does not exist."
        # 9. Missing Values
        elif "FIND_MISSING_VALUES" in decision_upper:
            result = find_missing_values()
        # 10. Unique Values
        elif "GET_UNIQUE_VALUES" in decision_upper:
            if columns_text == "NONE":
                result = "No column was provided."
            else:
                column = columns_text.split(",")[0].strip()
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = get_unique_values(
                        matching_columns[0]
                    )
                else:
                    result = f"Column '{column}' does not exist."
        # 11. Dataset Overview
        elif "DATASET_OVERVIEW" in decision_upper:
            result = dataset_overview()
        # 12. Unknown Tool
        else:
            result = (
                "The agent could not determine "
                "the appropriate tool."
            )
        # Print Tool Result
        print("Tool Result:")
        print(result)
        # Generate Natural Language Answer
        answer = generate_answer(
            user_question,
            result
        )
        print()
        print("Agent Answer:")
        print(answer)
        return answer
    # Error Handling
    except Exception as e:
        print("Agent Error:")
        print(str(e))
        user_message = handle_agent_error(e)
        print()
        print("Agent Response:")
        print(user_message)
        return user_message
```

```python
import matplotlib.pyplot as plt

def plot_histogram(column):
    """Plots a histogram for a numerical column."""
    if column not in df.columns:
        return f"Column '{column}' does not exist."
    if not pd.api.types.is_numeric_dtype(df[column]):
        return f"Column '{column}' is not numerical."
    plt.figure(figsize=(8, 5))
    plt.hist(df[column].dropna(), bins=20)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Distribution of {column}")
    plt.show()
    return f"Histogram created for {column}."
```

```python
import matplotlib.pyplot as plt
def plot_bar_chart(column):
    """Plots a bar chart for a categorical column."""
    if column not in df.columns:
        return f"Column '{column}' does not exist."
    if pd.api.types.is_numeric_dtype(df[column]):
        return f"Column '{column}' is numerical. Use a histogram instead."
    counts = df[column].value_counts()
    plt.figure(figsize=(8, 5))
    counts.plot(kind="bar")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.title(f"{column} Distribution")
    plt.xticks(rotation=45)
    plt.show()
    return f"Bar chart created for {column}."
```

```python
def calculate_correlation(column1, column2):
    """Calculates correlation between two numerical columns."""
    if column1 not in df.columns:
        return f"Column '{column1}' does not exist."
    if column2 not in df.columns:
        return f"Column '{column2}' does not exist."
    if not pd.api.types.is_numeric_dtype(df[column1]):
        return f"Column '{column1}' is not numerical."
    if not pd.api.types.is_numeric_dtype(df[column2]):
        return f"Column '{column2}' is not numerical."
    correlation = df[column1].corr(df[column2])
    return {
        "column_1": column1,
        "column_2": column2,
        "correlation": round(correlation, 4)
    }
```

```python
def plot_scatter(column1, column2):
    """Creates a scatter plot between two numerical columns."""

    if column1 not in df.columns:
        return f"Column '{column1}' does not exist."

    if column2 not in df.columns:
        return f"Column '{column2}' does not exist."

    if not pd.api.types.is_numeric_dtype(df[column1]):
        return f"Column '{column1}' is not numerical."

    if not pd.api.types.is_numeric_dtype(df[column2]):
        return f"Column '{column2}' is not numerical."

    plt.figure(figsize=(8, 5))

    plt.scatter(
        df[column1],
        df[column2]
    )

    plt.xlabel(column1)
    plt.ylabel(column2)

    plt.title(f"{column1} vs {column2}")

    plt.show()

    return f"Scatter plot created for {column1} and {column2}."
```

```python
def generate_answer(user_question, tool_result):
    if isinstance(tool_result, dict):
        if "mean" in tool_result:
            return (
                f"For the requested column, the mean is "
                f"{tool_result['mean']:.2f}, the median is "
                f"{tool_result['median']:.2f}, the minimum is "
                f"{tool_result['minimum']:.2f}, and the maximum is "
                f"{tool_result['maximum']:.2f}."
            )
        if "correlation" in tool_result:
            correlation = tool_result["correlation"]
            if correlation > 0.7:
                strength = "strong positive"
            elif correlation > 0.3:
                strength = "moderate positive"
            elif correlation > -0.3:
                strength = "weak or no"
            elif correlation > -0.7:
                strength = "moderate negative"
            else:
                strength = "strong negative"
            return (
                f"The correlation between "
                f"{tool_result['column_1']} and "
                f"{tool_result['column_2']} is "
                f"{correlation}. This indicates a "
                f"{strength} relationship."
            )
        if "rows" in tool_result and "columns" in tool_result:
            return (
                f"The dataset contains "
                f"{tool_result['rows']} rows and "
                f"{tool_result['columns']} columns."
            )
    return str(tool_result)
```

```python
def handle_agent_error(error):
    error_message = str(error)
    if "429" in error_message or "quota" in error_message.lower():
        return (
            "Gemini API quota has been exceeded. "
            "Please wait for the quota to reset before trying another AI request."
        )
    elif "404" in error_message:
        return (
            "The requested Gemini model or API endpoint was not found."
        )
    elif "401" in error_message or "403" in error_message:
        return (
            "There is a problem with the Gemini API key or its permissions."
        )
    elif "column" in error_message.lower():
        return (
            "There was a problem with the selected dataset column. "
            "Please check the column name."
        )
    else:
        return (
            f"Something went wrong while running the agent: {error_message}"
        )
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
from google import genai
import os
# PAGE CONFIGURATION
st.set_page_config(
    page_title="AI Data Analysis Agent",
    page_icon="🤖",
    layout="wide"
)
# GEMINI API
API_KEY = os.getenv("GEMINI_API_KEY")
if not API_KEY:
    st.error("Gemini API key not found.")
    st.info("Please set GEMINI_API_KEY before starting Streamlit.")
    st.stop()
client = genai.Client(api_key=API_KEY)
# TITLE
st.title("🤖 AI Data Analysis Agent")
st.write(
    "Upload a CSV file and ask questions about your dataset. "
    "The AI agent will select the appropriate analysis tool."
)
# CSV UPLOAD
uploaded_file = st.file_uploader(
    "Upload your CSV file",
    type=["csv"]
)
if uploaded_file is not None:
    df = pd.read_csv(uploaded_file)
    st.success("Dataset uploaded successfully!")
    # DATASET PREVIEW
    st.subheader("📊 Dataset Preview")
    st.dataframe(df.head())
    # DATASET INFORMATION
    st.subheader("📋 Dataset Information")
    col1, col2 = st.columns(2)
    with col1:
        st.metric("Rows", df.shape[0])
    with col2:
        st.metric("Columns", df.shape[1])
    # TOOL 1 — INSPECT DATASET
    def inspect_dataset():
        return {
            "rows": df.shape[0],
            "columns": df.shape[1],
            "column_names": df.columns.tolist(),
            "data_types": df.dtypes.astype(str).to_dict(),
            "missing_values": df.isnull().sum().to_dict()
        }
    # TOOL 2 — SINGLE STATISTICS
    def calculate_statistics(column):
        if column not in df.columns:
            return f"Column '{column}' does not exist."
        if not pd.api.types.is_numeric_dtype(df[column]):
            return f"Column '{column}' is not numerical."
        return {
            "column": column,
            "mean": df[column].mean(),
            "median": df[column].median(),
            "minimum": df[column].min(),
            "maximum": df[column].max(),
            "standard_deviation": df[column].std()
        }
    # TOOL 3 — MULTIPLE STATISTICS
    def calculate_multiple_statistics(columns):
        results = {}
        for column in columns:
            if column not in df.columns:
                results[column] = (
                    f"Column '{column}' does not exist."
                )
                continue
            if not pd.api.types.is_numeric_dtype(df[column]):
                results[column] = (
                    f"Column '{column}' is not numerical."
                )
                continue
            results[column] = {
                "mean": df[column].mean(),
                "median": df[column].median(),
                "minimum": df[column].min(),
                "maximum": df[column].max(),
                "standard_deviation": df[column].std()
            }
        return results
    # TOOL 4 — MISSING VALUES
    def find_missing_values():
        missing = df.isnull().sum()
        missing = missing[missing > 0]
        if len(missing) == 0:
            return "There are no missing values in the dataset."
        return missing.to_dict()
    # TOOL 5 — UNIQUE VALUES
    def get_unique_values(column):
        if column not in df.columns:
            return f"Column '{column}' does not exist."
        return df[column].unique().tolist()
    # TOOL 6 — DATASET OVERVIEW
    def dataset_overview():
        numerical_columns = (
            df.select_dtypes(include="number")
            .columns
            .tolist()
        )
        categorical_columns = (
            df.select_dtypes(exclude="number")
            .columns
            .tolist()
        )
        return {
            "rows": df.shape[0],
            "columns": df.shape[1],
            "numerical_columns": numerical_columns,
            "categorical_columns": categorical_columns,
            "missing_values": df.isnull().sum().to_dict(),
            "duplicate_rows": int(df.duplicated().sum())
        }
    # TOOL 7 — HISTOGRAM
    def plot_histogram(column):
        if column not in df.columns:
            return f"Column '{column}' does not exist."
        if not pd.api.types.is_numeric_dtype(df[column]):
            return f"Column '{column}' is not numerical."
        fig, ax = plt.subplots(figsize=(8, 5))
        ax.hist(
            df[column].dropna(),
            bins=20
        )
        ax.set_xlabel(column)
        ax.set_ylabel("Frequency")
        ax.set_title(
            f"Distribution of {column}"
        )
        st.pyplot(fig)
        plt.close(fig)
        return f"Histogram created for {column}."
    # TOOL 8 — BAR CHART
    def plot_bar_chart(column):
        if column not in df.columns:
            return f"Column '{column}' does not exist."
        if pd.api.types.is_numeric_dtype(df[column]):
            return (
                f"Column '{column}' is numerical. "
                "Use a histogram instead."
            )
        counts = df[column].value_counts()
        fig, ax = plt.subplots(figsize=(8, 5))
        counts.plot(
            kind="bar",
            ax=ax
        )
        ax.set_xlabel(column)
        ax.set_ylabel("Count")
        ax.set_title(
            f"{column} Distribution"
        )
        plt.xticks(rotation=45)
        st.pyplot(fig)
        plt.close(fig)
        return f"Bar chart created for {column}."
    # TOOL 9 — CORRELATION
    def calculate_correlation(column1, column2):
        if column1 not in df.columns:
            return f"Column '{column1}' does not exist."
        if column2 not in df.columns:
            return f"Column '{column2}' does not exist."
        if not pd.api.types.is_numeric_dtype(df[column1]):
            return f"Column '{column1}' is not numerical."
        if not pd.api.types.is_numeric_dtype(df[column2]):
            return f"Column '{column2}' is not numerical."
        correlation = df[column1].corr(
            df[column2]
        )
        return {
            "column_1": column1,
            "column_2": column2,
            "correlation": round(
                correlation,
                4
            )
        }
    # TOOL 10 — SCATTER PLOT
    def plot_scatter(column1, column2):
        if column1 not in df.columns:
            return f"Column '{column1}' does not exist."
        if column2 not in df.columns:
            return f"Column '{column2}' does not exist."
        if not pd.api.types.is_numeric_dtype(df[column1]):
            return f"Column '{column1}' is not numerical."
        if not pd.api.types.is_numeric_dtype(df[column2]):
            return f"Column '{column2}' is not numerical."
        fig, ax = plt.subplots(figsize=(8, 5))
        ax.scatter(
            df[column1],
            df[column2]
        )
        ax.set_xlabel(column1)
        ax.set_ylabel(column2)
        ax.set_title(
            f"{column1} vs {column2}"
        )
        st.pyplot(fig)
        plt.close(fig)
        return (
            f"Scatter plot created for "
            f"{column1} and {column2}."
        )
    # AI AGENT DECISION
    def agent_decide(user_question):
        tools_description = """

Available tools:

1. inspect_dataset
   Use for basic dataset structure.

2. calculate_statistics
   Use for statistics of ONE numerical column.

3. calculate_multiple_statistics
   Use for statistics of TWO OR MORE numerical columns.

4. find_missing_values
   Use for missing or null values.

5. get_unique_values
   Use for unique categories.

6. dataset_overview
   Use for complete dataset summary.

7. plot_histogram
   Use for histogram of ONE numerical column.

8. plot_bar_chart
   Use for bar chart of ONE categorical column.

9. calculate_correlation
   Use for correlation between TWO numerical columns.

10. plot_scatter
    Use for scatter plot between TWO numerical columns.
"""

        prompt = f"""

You are an AI Data Analysis Agent.

{tools_description}

User question:

{user_question}

Choose the ONE most appropriate tool.

Rules:

- Numerical correlation value:
  calculate_correlation

- Graph or visualization of two numerical columns:
  plot_scatter

- Both correlation and scatter plot:
  calculate_correlation

- Statistics of one numerical column:
  calculate_statistics

- Statistics of multiple numerical columns:
  calculate_multiple_statistics

Return EXACTLY:

TOOL: tool_name
COLUMNS: column1, column2

If one column is required:

COLUMNS: column1

If no column is required:

COLUMNS: NONE
"""
        response = client.interactions.create(
            model="gemini-3.6-flash",
            input=prompt
        )
        return response.output_text
    # NATURAL LANGUAGE ANSWER
    def generate_answer(
        user_question,
        tool_result
    ):
        if isinstance(tool_result, dict):
            if "mean" in tool_result:
                return (
                    f"For {tool_result.get('column', 'the requested column')}, "
                    f"the mean is "
                    f"{tool_result['mean']:.2f}, "
                    f"the median is "
                    f"{tool_result['median']:.2f}, "
                    f"the minimum is "
                    f"{tool_result['minimum']:.2f}, "
                    f"and the maximum is "
                    f"{tool_result['maximum']:.2f}."
                )
            if "correlation" in tool_result:

                correlation = tool_result[
                    "correlation"
                ]
                if correlation > 0.7:
                    strength = "strong positive"
                elif correlation > 0.3:
                    strength = "moderate positive"
                elif correlation > -0.3:
                    strength = "weak or no"
                elif correlation > -0.7:
                    strength = "moderate negative"
                else:
                    strength = "strong negative"
                return (
                    f"The correlation between "
                    f"{tool_result['column_1']} and "
                    f"{tool_result['column_2']} is "
                    f"{correlation}. "
                    f"This indicates a "
                    f"{strength} relationship."
                )
            if (
                "rows" in tool_result
                and "columns" in tool_result
            ):
                return (
                    f"The dataset contains "
                    f"{tool_result['rows']} rows "
                    f"and "
                    f"{tool_result['columns']} columns."
                )
        return str(tool_result)
    # ERROR HANDLING
    def handle_agent_error(error):
        error_message = str(error)
        if (
            "429" in error_message
            or "quota" in error_message.lower()
        ):
            return (
                "Gemini API quota has been exceeded. "
                "Please wait for the quota to reset "
                "before trying another AI request."
            )
        elif "404" in error_message:
            return (
                "The requested Gemini model or "
                "API endpoint was not found."
            )
        elif (
            "401" in error_message
            or "403" in error_message
        ):
            return (
                "There is a problem with the Gemini "
                "API key or its permissions."
            )
        elif "column" in error_message.lower():
            return (
                "There was a problem with the selected "
                "dataset column. Please check the "
                "column name."
            )
        else:
            return (
                "Something went wrong while running "
                f"the agent: {error_message}"
            )
    # RUN AGENT
    def run_agent(user_question):
        try:
            # Ask Gemini to choose tool
            decision = agent_decide(
                user_question
            )
            st.subheader("🤖 Agent Decision")
            st.code(decision)
            decision_upper = decision.upper()
            if "COLUMNS:" in decision_upper:
                columns_text = (
                    decision_upper
                    .split("COLUMNS:", 1)[1]
                    .strip()
                )
            else:
                columns_text = "NONE"
            # SCATTER PLOT
            if "PLOT_SCATTER" in decision_upper:
                columns = [
                    col.strip()
                    for col in columns_text.split(",")
                ]
                matching_columns = []
                for column in columns[:2]:
                    matches = [
                        real_col
                        for real_col in df.columns
                        if real_col.upper() == column
                    ]
                    if matches:
                        matching_columns.append(
                            matches[0]
                        )
                if len(matching_columns) == 2:
                    result = plot_scatter(
                        matching_columns[0],
                        matching_columns[1]
                    )
                else:
                    result = (
                        "One or both columns "
                        "do not exist."
                    )
            # CORRELATION
            elif (
                "CALCULATE_CORRELATION"
                in decision_upper
            ):
                columns = [
                    col.strip()
                    for col in columns_text.split(",")
                ]
                matching_columns = []
                for column in columns[:2]:
                    matches = [
                        real_col
                        for real_col in df.columns
                        if real_col.upper() == column
                    ]
                    if matches:
                        matching_columns.append(
                            matches[0]
                        )
                if len(matching_columns) == 2:
                    result = calculate_correlation(
                        matching_columns[0],
                        matching_columns[1]
                    )
                else:
                    result = (
                        "One or both columns "
                        "do not exist."
                    )
            # HISTOGRAM
            elif "PLOT_HISTOGRAM" in decision_upper:
                column = (
                    columns_text
                    .split(",")[0]
                    .strip()
                )
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = plot_histogram(
                        matching_columns[0]
                    )
                else:
                    result = (
                        f"Column '{column}' "
                        "does not exist."
                    )
            # BAR CHART
            elif "PLOT_BAR_CHART" in decision_upper:
                column = (
                    columns_text
                    .split(",")[0]
                    .strip()
                )
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = plot_bar_chart(
                        matching_columns[0]
                    )
                else:
                    result = (
                        f"Column '{column}' "
                        "does not exist."
                    )
            # INSPECT DATASET
            elif "INSPECT_DATASET" in decision_upper:
                result = inspect_dataset()
            # MULTIPLE STATISTICS
            elif (
                "CALCULATE_MULTIPLE_STATISTICS"
                in decision_upper
            ):
                columns = [
                    col.strip()
                    for col in columns_text.split(",")
                ]
                actual_columns = []
                for column in columns:
                    matches = [
                        real_col
                        for real_col in df.columns
                        if real_col.upper() == column
                    ]
                    if matches:
                        actual_columns.append(
                            matches[0]
                        )
                if actual_columns:
                    result = (
                        calculate_multiple_statistics(
                            actual_columns
                        )
                    )
                else:
                    result = (
                        "None of the requested "
                        "columns exist."
                    )
            # SINGLE STATISTICS
            elif (
                "CALCULATE_STATISTICS"
                in decision_upper
            ):
                column = (
                    columns_text
                    .split(",")[0]
                    .strip()
                )
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = calculate_statistics(
                        matching_columns[0]
                    )
                else:
                    result = (
                        f"Column '{column}' "
                        "does not exist."
                    )
            # MISSING VALUES
            elif (
                "FIND_MISSING_VALUES"
                in decision_upper
            ):
                result = find_missing_values()
            # UNIQUE VALUES
            elif (
                "GET_UNIQUE_VALUES"
                in decision_upper
            ):
                column = (
                    columns_text
                    .split(",")[0]
                    .strip()
                )
                matching_columns = [
                    col
                    for col in df.columns
                    if col.upper() == column
                ]
                if matching_columns:
                    result = get_unique_values(
                        matching_columns[0]
                    )
                else:
                    result = (
                        f"Column '{column}' "
                        "does not exist."
                    )
            # DATASET OVERVIEW
            elif (
                "DATASET_OVERVIEW"
                in decision_upper
            ):
                result = dataset_overview()
            else:
                result = (
                    "The agent could not determine "
                    "the appropriate tool."
                )
            # TOOL RESULT
            st.subheader("🔧 Tool Result")
            st.write(result)
            # FINAL ANSWER
            answer = generate_answer(
                user_question,
                result
            )
            st.subheader("💬 Agent Answer")
            st.success(answer)
        except Exception as e:
            error_message = handle_agent_error(e)
            st.error(error_message)
    # USER QUESTION
    st.divider()
    st.subheader("💬 Ask the Data Analysis Agent")
    user_question = st.text_input(
        "Enter your question:"
    )
    # ANALYZE BUTTON
    if st.button(
        "🔍 Analyze",
        type="primary"
    ):
        if user_question.strip():
            run_agent(
                user_question
            )
        else:
            st.warning(
                "Please enter a question."
            )
```
